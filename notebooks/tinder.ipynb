# Speed Dating with Tinder — Analyse Exploratoire

Données issues d'expériences de speed dating (2002–2004). Chaque ligne représente un échange de 4 minutes entre deux participants. L'objectif : comprendre ce qui pousse deux personnes à vouloir se revoir.

**5 questions à explorer :**
1. Quels sont les attributs les moins désirables selon le genre ?
2. L'attractivité est-elle aussi décisive qu'on le croit ?
3. Les intérêts partagés comptent-ils plus que l'origine raciale ?
4. Les participants se connaissent-ils eux-mêmes ?
5. Vaut-il mieux passer en début ou en fin de soirée ?

## Setup

```python
import pandas as pd
import plotly.express as px

pd.set_option('display.max_columns', None)
```

## Chargement des données

```python
df = pd.read_csv("../data/raw/Speed_Dating_Data.csv", encoding='latin-1')
print(f"{df.shape[0]} speed dates — {df.shape[1]} variables")
df.head()
```

```python
df.describe(include='all')
```

### Préparation des labels et normalisation

```python
# Labels lisibles
df['gender'] = df['gender'].replace({1: 'Hommes', 0: 'Femmes'})
df['match_label'] = df['match'].map({0: 'No Match', 1: 'Match'})
df['race_label'] = df['race'].map({
    1: 'Black/African Am.', 2: 'Européen/Caucasien',
    3: 'Latino/Hispanic',   4: 'Asiatique',
    5: 'Natif Américain',   6: 'Autre'
})
df['goal_label'] = df['goal'].map({
    1: "Se divertir",           2: "Rencontrer des gens",
    3: "Obtenir un RDV",        4: "Relation sérieuse",
    5: "Pour l'expérience",     6: "Autre"
})

# Normalisation des critères de sélection sur 100 points
attr_cols = ['attr1_1', 'sinc1_1', 'intel1_1', 'fun1_1', 'amb1_1', 'shar1_1']

def normalize_row(row):
    total = row[attr_cols].sum()
    if total == 0:
        return row[attr_cols]
    return (row[attr_cols] / total) * 100

df[attr_cols] = df[attr_cols].apply(normalize_row, axis=1)
```

```python
df.head()
```

## Qualité des données — Valeurs manquantes

Beaucoup de colonnes proviennent de questionnaires de suivi (J+1, J+30) qui n'ont pas tous été complétés.
On se concentre sur les variables du soir J, globalement bien renseignées.

```python
missing_data = df.isnull().mean().sort_values(ascending=False).reset_index()
missing_data.columns = ['Variable', 'Pct_manquants']
missing_data = missing_data[missing_data['Pct_manquants'] > 0]

fig = px.bar(
    missing_data, x='Pct_manquants', y='Variable', orientation='h',
    title="Pourcentage de valeurs manquantes par variable",
    color='Pct_manquants', color_continuous_scale='Reds', height=1200
)
fig.show()
```

Avant d'analyser les matchs, on dresse un portrait rapide de ceux qui ont participé à ces soirées.

```python
df_sun = df.groupby(['goal_label', 'match_label']).size().reset_index(name='nombre')

fig = px.sunburst(
    df_sun, path=['goal_label', 'match_label'], values='nombre',
    color='goal_label',
    title='Objectif de la soirée et résultat obtenu',
    width=750, height=750
)
fig.update_traces(textinfo='label+percent parent', insidetextorientation='radial')
fig.show()
```

## 2. Notre variable cible : le match

Un match se produit uniquement quand **les deux** participants disent "oui". C'est un événement rare.

```python
match_counts = df['match'].value_counts().reset_index()
match_counts['label'] = match_counts['match'].map({0: 'No Match', 1: 'Match'})

fig = px.pie(
    match_counts, values='count', names='label',
    title=f"Taux de match global — {df['match'].mean()*100:.1f}% des speed dates aboutissent à un match",
    color='label',
    color_discrete_map={'Match': '#2ecc71', 'No Match': '#e74c3c'},
    width=600, height=400
)
fig.show()
```

```python
taux_genre = df.groupby('gender')['match'].mean().reset_index()
taux_genre.columns = ['Genre', 'Taux de match']
taux_genre['Taux de match'] = (taux_genre['Taux de match'] * 100).round(1)

fig = px.bar(taux_genre, x='Genre', y='Taux de match',
             title='Taux de match par genre (%)',
             text='Taux de match',
             color='Genre',
             width=600, height=400)
fig.update_traces(texttemplate='%{text}%', textposition='outside')
fig.update_layout(showlegend=False)
fig.show()
```

**Interprétation :** Seulement ~16% des speed dates se terminent par un match. Hommes et femmes ont exactement le même taux — logique, puisqu'un match nécessite l'accord des deux. Ce taux de 16% sera notre référence tout au long de l'analyse.

## Question 1 — Qu'est-ce que chaque genre recherche, et qu'est-ce qu'il ne veut pas ?

Les participants répartissaient 100 points entre 6 critères selon leur importance dans le choix d'un partenaire. On regarde comment cette répartition diffère selon le genre.

```python
attr_labels = {
    'attr1_1': 'Attractivité', 'sinc1_1': 'Sincérité',  'intel1_1': 'Intelligence',
    'fun1_1':  'Fun',           'amb1_1':  'Ambition',   'shar1_1':  'Intérêts communs'
}

attr_by_gender = df.groupby('gender')[attr_cols].mean()
attr_by_gender.columns = [attr_labels[c] for c in attr_cols]

attr_long = attr_by_gender.reset_index().melt(
    id_vars='gender', var_name='Critère', value_name='Poids moyen'
)

fig = px.bar(
    attr_long, x='Critère', y='Poids moyen', color='gender', barmode='group',
    title='Ce que chaque genre recherche chez un partenaire (sur 100 points)',
    labels={'Poids moyen': 'Poids moyen (pts)', 'Critère': '', 'gender': 'Genre'},
    color_discrete_sequence=['#636EFA', '#EF553B'],
    width=800, height=500
)
fig.show()
```

**Interprétation :** Les hommes placent l'attractivité physique loin devant (~27 pts), tandis que les femmes répartissent plus équitablement — sincérité, intelligence et fun arrivent presque au même niveau. **L'attribut le moins valorisé par les deux genres est l'ambition**, ce qui suggère qu'elle pèse peu lors d'un premier contact de 4 minutes. Pour les hommes, c'est aussi la sincérité qui est sous-cotée comparée aux femmes.

## Question 2 — L'attractivité est-elle aussi déterminante qu'on le croit ?

Les participants lui accordent beaucoup d'importance. Mais est-ce vraiment ce qui fait la différence entre un match et un non-match dans la pratique ?

```python
# Écart de score entre Match et No Match pour chaque attribut
attributs = ['attr_o', 'sinc_o', 'intel_o', 'fun_o', 'amb_o', 'shar_o']
attr_names = {
    'attr_o': 'Attractivité', 'sinc_o': 'Sincérité',    'intel_o': 'Intelligence',
    'fun_o':  'Fun',           'amb_o':  'Ambition',     'shar_o':  'Intérêts partagés'
}

comp = df.groupby('match')[attributs].mean().T
comp.columns = ['No Match', 'Match']
comp['Différence'] = comp['Match'] - comp['No Match']
comp = comp.reset_index()
comp['index'] = comp['index'].map(attr_names)
comp = comp.sort_values('Différence', ascending=False)

fig = px.bar(
    comp, x='index', y='Différence',
    title="Écart de score entre Match et No Match — impact réel de chaque attribut",
    labels={'index': 'Attribut', 'Différence': 'Différence (Match − No Match)'},
    color='Différence', color_continuous_scale='Blues',
    width=750, height=500
)
fig.show()
```

```python
# Le score "j'ai apprécié cette personne" (like_o) : prédicteur global
like_comp = df.groupby(['gender', 'match_label'])['like_o'].mean().reset_index()

fig = px.bar(
    like_comp, x='gender', y='like_o', color='match_label', barmode='group',
    title="'J'ai apprécié cette personne' — score moyen selon le résultat",
    labels={'like_o': 'Score moyen (sur 10)', 'gender': 'Genre', 'match_label': 'Résultat'},
    color_discrete_map={'Match': '#2ecc71', 'No Match': '#e74c3c'},
    width=700, height=450
)
fig.show()
```

**Interprétation :** L'attractivité a bien l'impact le plus fort sur le match (écart ~1,5 pt), suivie par le fun et les intérêts partagés. La sincérité et l'ambition ont l'effet le plus faible. Mais le vrai révélateur est le score global "j'ai apprécié cette personne" (`like_o`) : il passe de ~4,5 en non-match à ~7,5 en match. **L'alchimie perçue globalement prédit mieux le match que n'importe quel critère isolé**, y compris l'attractivité.

## Question 3 — Les intérêts partagés comptent-ils plus que l'origine raciale ?

```python
# Corrélation d'intérêts calculée objectivement par les chercheurs (int_corr)
int_corr_comp = df.groupby('match_label')['int_corr'].mean().reset_index()

fig = px.bar(
    int_corr_comp, x='match_label', y='int_corr',
    title='Corrélation des intérêts entre participants selon le résultat',
    labels={'int_corr': 'Corrélation moyenne des intérêts', 'match_label': 'Résultat'},
    color='match_label',
    color_discrete_map={'Match': '#2ecc71', 'No Match': '#e74c3c'},
    width=600, height=400
)
fig.show()
```

```python
# Comparaison directe des taux de match
taux_global = df['match'].mean() * 100

comparison_df = pd.DataFrame({
    'Facteur': ['Intérêts partagés
élevés (score ≥ 8)', 'Même origine
raciale'],
    'Taux de match (%)': [
        round(df[df['shar_o'] >= 8]['match'].mean() * 100, 1),
        round(df[df['samerace'] == 1]['match'].mean() * 100, 1)
    ]
})

fig = px.bar(
    comparison_df, x='Facteur', y='Taux de match (%)',
    title='Taux de match : intérêts partagés vs même origine raciale',
    text='Taux de match (%)',
    color='Facteur',
    width=650, height=450
)
fig.update_traces(texttemplate='%{text}%', textposition='outside')
fig.add_hline(y=taux_global, line_dash='dash', line_color='grey',
              annotation_text=f"Moyenne globale : {taux_global:.1f}%",
              annotation_position='top right')
fig.show()
```

**Interprétation :** La réponse est sans appel. Des intérêts partagés élevés (≥ 8/10) **doublent presque le taux de match** (~34% vs 16% en moyenne). En revanche, partager la même origine raciale n'apporte qu'un effet marginal (~17%, à peine au-dessus de la moyenne). La corrélation d'intérêts calculée objectivement confirme la même tendance. **Ce qu'on aime faire rapproche bien plus que d'où on vient.**

## Question 4 — Les participants se connaissent-ils eux-mêmes ?

Savent-ils comment les autres les perçoivent ? Et peuvent-ils prédire si ça va matcher ?

```python
# Auto-évaluation de l'attractivité vs note réellement reçue
self_vs_real = df[['gender', 'attr3_1', 'attr_o']].dropna()

fig = px.scatter(
    self_vs_real, x='attr3_1', y='attr_o', color='gender',
    trendline='ols',
    title='Auto-évaluation vs note réellement reçue (attractivité)',
    labels={
        'attr3_1': 'Comment je me note (sur 10)',
        'attr_o':  'Comment les autres me notent (sur 10)',
        'gender':  'Genre'
    },
    opacity=0.25, width=700, height=500
)
# Ligne d'alignement parfait
fig.add_shape(type='line', x0=1, y0=1, x1=10, y1=10,
              line=dict(color='black', dash='dot', width=1))
fig.add_annotation(x=8, y=6.5, text="Alignement parfait →", showarrow=False,
                   font=dict(size=10, color='black'))
fig.show()
```

```python
# Prédiction du match : est-ce que le partenaire "sent" le match venir ?
prob_comp = df.groupby('match_label')['prob_o'].mean().reset_index()

fig = px.bar(
    prob_comp, x='match_label', y='prob_o',
    title='Probabilité de match estimée par le partenaire vs résultat réel',
    labels={'prob_o': 'Probabilité estimée (sur 10)', 'match_label': 'Résultat réel'},
    color='match_label',
    color_discrete_map={'Match': '#2ecc71', 'No Match': '#e74c3c'},
    text='prob_o',
    width=600, height=400
)
fig.update_traces(texttemplate='%{text:.1f}', textposition='outside')
fig.show()
```

**Interprétation :** Le nuage de points se situe systématiquement **au-dessus** de la diagonale — les participants se surévaluent presque tous. Ce biais de positivité est présent chez les deux genres, légèrement plus marqué chez les hommes. Sur la prédiction, les participants qui ont obtenu un match s'y attendaient un peu plus (~5/10 vs ~3/10). Ils "sentent" quelque chose, mais l'écart reste faible : **on est de mauvais juges de notre propre valeur sur le marché de la séduction.**

## Question 5 — Vaut-il mieux être le premier ou le dernier speed date de la nuit ?

```python
match_by_order = df.groupby('order')['match'].mean().reset_index()
match_by_order.columns = ['Position', 'Taux de match']
match_by_order['Taux de match'] = (match_by_order['Taux de match'] * 100).round(2)

fig = px.line(
    match_by_order, x='Position', y='Taux de match',
    title='Taux de match (%) selon la position dans la soirée',
    markers=True,
    labels={'Taux de match': 'Taux de match (%)'},
    width=750, height=480
)
fig.add_hline(
    y=df['match'].mean() * 100,
    line_dash='dash', line_color='red',
    annotation_text=f"Moyenne : {df['match'].mean()*100:.1f}%",
    annotation_position='top right'
)
fig.show()
```

**Interprétation :** Pas de tendance linéaire claire — ni les premiers ni les derniers ne sont systématiquement avantagés. On observe des variations autour de la moyenne sans pattern stable. L'effet de position est **minoritaire comparé aux critères personnels**. Si un conseil pratique est possible : éviter le milieu de soirée (positions 8–12) où le taux semble légèrement sous la moyenne, sans doute un creux d'attention.

## Conclusion

Cette analyse de 8 378 speed dates permet de répondre aux 5 questions initiales :

| # | Question | Réponse |
|---|---|---|
| 1 | Attributs les moins désirables | L'**ambition** est le critère le moins valorisé par les deux genres. Chez les hommes, la sincérité passe aussi loin derrière l'attractivité. |
| 2 | Attractivité vs réalité | Elle a bien le plus fort impact, mais c'est le **score global d'appréciation** (`like_o`) qui prédit le mieux un match. L'alchimie prime. |
| 3 | Intérêts partagés vs origine | Les **intérêts communs** ont un impact deux fois supérieur sur le taux de match (~34%) vs même origine (~17%). |
| 4 | Auto-perception | Les participants **surestiment** leur attractivité. Ils "sentent" vaguement un match possible mais restent de mauvais juges. |
| 5 | Position dans la soirée | L'effet est **limité**. Aucun slot n'est clairement avantageux. |
